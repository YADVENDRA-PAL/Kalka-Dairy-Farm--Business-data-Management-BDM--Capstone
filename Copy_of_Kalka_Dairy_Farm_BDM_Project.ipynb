```python
milk = pd.read_excel(file, sheet_name="Daily_Milk_Production")
sales = pd.read_excel(file, sheet_name="Milk_Sales")
expenses = pd.read_excel(file, sheet_name="Expenses")
feed = pd.read_excel(file, sheet_name="Feed_Consumption")
customers = pd.read_excel(file, sheet_name="Customers")
animal = pd.read_excel(file, sheet_name="Animal_Master")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert Date column safely
milk["Date"] = pd.to_datetime(
    milk["Date"],
    dayfirst=True,
    format="mixed",
    errors="coerce"
)

# Remove rows where Date could not be converted
milk = milk.dropna(subset=["Date"])

# Sort by date
milk = milk.sort_values("Date")

# Group by Date and calculate total milk production
daily = milk.groupby("Date")["Total Milk (L)"].sum()

# Plot graph
plt.figure(figsize=(12,5))
plt.plot(daily.index, daily.values, marker='o', linewidth=2)
plt.title("Daily Milk Production")
plt.xlabel("Date")
plt.ylabel("Total Milk (Litres)")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
animal_prod = milk.groupby("Animal Name")["Total Milk (L)"].sum()

plt.figure(figsize=(10,5))
plt.bar(animal_prod.index, animal_prod.values)
plt.title("Animal-wise Milk Production")
plt.xlabel("Animal")
plt.ylabel("Milk (Litres)")
plt.xticks(rotation=45)
plt.show()
```

```python
milk["Month"] = milk["Date"].dt.month_name()

monthly = milk.groupby("Month")["Total Milk (L)"].sum()

months=["January","February","March","April","May","June"]

monthly=monthly.reindex(months)

plt.figure(figsize=(10,5))
plt.bar(monthly.index, monthly.values)
plt.title("Monthly Milk Production")
plt.ylabel("Milk (Litres)")
plt.show()
```

```python
sales["Date"] = pd.to_datetime(sales["Date"])

sales["Month"] = sales["Date"].dt.month_name()

revenue = sales.groupby("Month")["Amount (₹)"].sum()

revenue = revenue.reindex(months)

plt.figure(figsize=(10,5))
plt.plot(revenue.index,revenue.values,marker='o')
plt.title("Monthly Revenue")
plt.ylabel("Revenue (₹)")
plt.grid(True)
plt.show()
```

```python
expense = expenses.groupby("Category")["Amount (₹)"].sum()

plt.figure(figsize=(7,7))
plt.pie(expense.values,
        labels=expense.index,
        autopct="%1.1f%%")
plt.title("Expense Distribution")
plt.show()
```

```python
feed["Date"] = pd.to_datetime(feed["Date"])

feed["Month"] = feed["Date"].dt.month_name()

feed_month = feed.groupby("Month")["Quantity (Kg)"].sum()

feed_month = feed_month.reindex(months)

plt.figure(figsize=(10,5))
plt.bar(feed_month.index,feed_month.values)
plt.title("Monthly Feed Consumption")
plt.ylabel("Kilograms")
plt.show()
```

```python
customer = customers["Customer Type"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(customer.values,
        labels=customer.index,
        autopct="%1.1f%%")
plt.title("Customer Type")
plt.show()
```

```python
status = animal["Status"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(status.values,
        labels=status.index,
        autopct="%1.1f%%")
plt.title("Animal Status")
plt.show()
```

```python
top = sales.groupby("Customer Name")["Amount (₹)"].sum()

top = top.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top.index,top.values)
plt.xticks(rotation=60)
plt.title("Top 10 Customers")
plt.ylabel("Revenue (₹)")
plt.show()
```
